**isolation forest anomaly detection**

```python
import pandas as pd 
df=pd.read_csv("TestAnomalyUnlabled.csv")
df.head()
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.iloc[:,1],df.iloc[:,2],1.5,linewidths=0.3)
```

```python
from sklearn.ensemble import IsolationForest
```

```python
dp=(df.iloc[:,1],df.iloc[:,2])
```

```python
clf=IsolationForest(contamination='auto')
clf.fit(df[['temperature_c',"humidity_pct","pressure_hpa"]])
prediction=clf.predict(df[['temperature_c',"humidity_pct","pressure_hpa"]])
```

```python
prediction
```

```python
import numpy as np
index=np.where(prediction<0)
index
```

```python
x=df.values
```

```python

plt.scatter(df.iloc[:,1],df.iloc[:,2],0.5)
plt.scatter(x[index,1],x[index,2],0.8,edgecolors="r")
```
