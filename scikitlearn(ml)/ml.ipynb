```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from  sklearn.preprocessing import PolynomialFeatures
from sklearn.ensemble import(
    HistGradientBoostingRegressor,
    RandomForestRegressor
)
import joblib
```

```python
housing=datasets.fetch_california_housing()
print(housing.feature_names)
```

```python
x=housing.data
y=housing.target



#classifying data into test sample n train sample
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=432 #ts will ensure suffling is consitent
) 
LR=LinearRegression()
GBR=HistGradientBoostingRegressor()
RFR=RandomForestRegressor(
         n_jobs=-1
)

for i in [LR,GBR,RFR]:
        i.fit(x_train,y_train)  # model will go sample by sample n compare each set of festures to its target
        y_pred=i.predict(x_test)
        r2 = r2_score(y_test, y_pred)
        print(i,r2)
#now we get base line n ml goal is to take this base line to more accuracy
```

finding best no of trees for HistGradientBoostingRegressor

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=432 #ts will ensure suffling is consitent
) 

#100 is defualt no of trees
for i in [100,250,200,250,300,350]:
        model=HistGradientBoostingRegressor(
                max_iter=i
        )
        model.fit(x_train,y_train)  # model will go sample by sample n compare each set of festures to its target
        y_pred=model.predict(x_test)
        r2 = r2_score(y_test, y_pred)
        print(i,r2)
```

using learning rate
hyperparatimization

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=432 #ts will ensure suffling is consitent
) 
for j in [0.1,0.05,0.001]:
 #100 is defualt no of trees
 for i in [200,250,300]:
        model=HistGradientBoostingRegressor(
                max_iter=i,
                learning_rate=j,
                random_state=432

        )
        model.fit(x_train,y_train)  # model will go sample by sample n compare each set of festures to its target
        y_pred=model.predict(x_test)
        r2 = r2_score(y_test, y_pred)
        print(i,j,r2)
```

saving it as lib soo we dont need to train again

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=432 #ts will ensure suffling is consitent
) 

model=HistGradientBoostingRegressor(
    max_iter=300,
    learning_rate=0.1,
    random_state=432
    
)
model.fit(x_train,y_train)  # model will go sample by sample n compare each set of festures to its target

joblib.dump(model,"my_model.joblib")

y_pred=model.predict(x_test)
r2 = r2_score(y_test, y_pred)
print(r2)
```

to use further anywhere 

```python
local_model=joblib.load("my_model.joblib")

y_pred=local_model.predict(x_test)
r2 = r2_score(y_test, y_pred)
print(r2)
```
