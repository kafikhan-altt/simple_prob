```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
penguins=sns.load_dataset('penguins')
penguins.head()
```

```python
penguins["species"].value_counts()
```

```python
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("darkgrid")
sns.set_context("notebook")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island",palette="Dark2")
```

```python
sns.scatterplot(data=penguins,x="species",y="body_mass_g",hue="island",style="sex")
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island",dodge=True)
```

```python
sns.set_context("talk")
sns.swarmplot(data=penguins,x="species",y="body_mass_g",hue="island")
```

```python
#histogram
sns.set_context("notebook")
sns.histplot(data=penguins,x="body_mass_g",hue="sex")
```

```python
#regression plot:
sns.regplot(data=penguins,x="body_mass_g",y="flipper_length_mm")
```

```python
#line plot
sns.lineplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",style="sex")
```

```python
#jointplots
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="sex")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",kind="reg")
```

```python
#bar plot
sns.barplot(data=penguins,x="species",y="body_mass_g",hue="sex",palette=["pink","blue"],estimator=np.var)
```

```python
sns.countplot(data=penguins,x="species",hue="island",palette="Set2")  
```

```python
#box plot
sns.boxplot(data=penguins,x="species",y="body_mass_g",hue="island")
```

```python
#voilet pint
sns.violinplot(data=penguins,x="species",y="body_mass_g",hue="island")
```

```python
sns.violinplot(data=penguins,x="species",y="body_mass_g",palette="pastel")
sns.swarmplot(data=penguins,x="species",y="body_mass_g",hue="island",size=3,palette="dark")
```

```python
#KDE
sns.kdeplot(data=penguins,x="body_mass_g",hue="species",fill=True)
```

```python
#heatmap
penguins.head()
```

```python
columns=["bill_length_mm"	,"bill_depth_mm",	"flipper_length_mm",	"body_mass_g"]
```

```python
penguins[columns].corr()
```

```python
sns.heatmap(data=penguins[columns].corr(),annot=True,cmap="coolwarm")
```

```python
#rug plots
sns.rugplot(data=penguins,x="body_mass_g",hue="species",palette="Set2")
```

```python
#pair plot
sns.pairplot(data= penguins,hue="sex",palette="Set2",diag_kind="kde")
```

```python
#pair grid
graph=sns.PairGrid(data=penguins,hue="sex",palette="Set2")
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
graph.add_legend()

       
```
